# Function

1.  What is the difference between a function and a method in Python?
- **Function :**

 A function is a block of code that performs a specific task and is not tied to any object.

 It is defined using the def keyword or lambda expression.

```python
# Funtion example

def fname(name):
    return f"Hello, {name}!"

fname("raju")
```

 - **Method :**
- A method is a function that is associated with an object, typically defined inside a class.

  - It is called on an instance of a class and usually has access to that instance through the self parameter.

```python
# python Method example

class Greeter:
    def greet(self, name):
        return f"Hello, {name}!"

g = Greeter()
g.greet("Raju")
```

4.  What is the purpose of the `return` statement in a Python function?
- The return statement in a Python function is used to :

  **Send a Result Back to the Caller**
  
  It exits the function and passes data back to the place where the function was called.
  

```python
# Example
def add(a, b):
    return a + b

result = add(3, 4)
```

- **End Function Execution**

 Once return is executed, the function immediately stops, and no further code in that function runs.

```python
# Example
def test():
    return "Done"
    print("This will never be printed")  # Unreachable
```

-  **Return Multiple Values**

 Python allows returning multiple values as a tuple

```python
def stats(a, b):
    return a + b, a * b

s, p = stats(2, 3)  # s = 5, p = 6
```

5. What are iterators in Python and how do they differ from iterables?
- Iterables

An iterable is any object you can loop over with a for loop.

Examples: lists, tuples, strings, dictionaries, sets, and custom objects that implement __iter__().

Key Property:

It has an __iter__() method that returns an iterator.


```python
my_list = [1, 2, 3]  # This is an iterable
```

- Iterators

An iterator is an object that produces items one at a time using the next() function.

It must implement two methods:

__iter__() → returns the iterator itself

__next__() → returns the next item, or raises StopIteration when done

```python
my_list = [1, 2, 3]
iterator = iter(my_list)  # Now it's an iterator

print(next(iterator))  # Output: 1
print(next(iterator))  # Output: 2
```

6.  Explain the concept of generators in Python and how they are defined.
- A generator is a special type of function that remembers where it left off and gives you one value at a time, only when you ask for it.

- It’s like a TV series — you watch one episode (value), and the next episode only plays when you ask for it (using next()).


```python
#  Use yield instead of return:
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1
```

```python
# How to Use It:
numbers = count_up_to(3)

print(next(numbers))  # prints 1
print(next(numbers))  # prints 2
print(next(numbers))  # prints 3
```

7. What are the advantages of using generators over regular functions?

- Generators in Python offer several important advantages over regular functions, especially when dealing with large data or streaming processes.
- Advantages of Using Generators
1. **Memory Efficiency**

 Generators don’t store all values in memory. They produce items one at a time, on the fly.

 Great for working with large datasets, such as reading big files or processing data streams.

```python
# Example
def generate_numbers():
    for i in range(1_000_000):
        yield i
```

- 2. **Lazy Evaluation**

 Values are calculated only when needed, not all at once.

 This means faster startup and less CPU usage if you don’t need every value.

- 4.**Clean Syntax for Iteration**

 You can write complex loops simply, without having to manage state yourself.


```python
# Example
def countdown(n):
    while n > 0:
        yield n
        n -= 1
```

8. What is a lambda function in Python and when is it typically used?
- A lambda function in Python is a small, anonymous function defined using the lambda keyword instead of def.

- What is a Lambda Function?

 It's a one-line function with no name.

 It can take any number of arguments, but can only contain a single expression.

 The result of the expression is automatically returned.


```python
# Basic Syntax:
lambda arguments: expression
```

```python
# Example
add = lambda x, y: x + y
print(add(2, 3))
```

- **When to Use Lambda Functions**

 Lambda functions are typically used for simple operations and often as short, throwaway functions, especially in:
1. Functions like map(), filter(), and reduce()


```python
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x*x, nums))  # [1, 4, 9, 16]
```

2. Sorting with custom keys

```python
words = ["apple", "banana", "cherry"]
words.sort(key=lambda word: len(word))  # Sort by length
```

3. Simple event handlers or callbacks

```python
button.on_click(lambda event: print("Clicked!"))
```

9. Explain the purpose and usage of the `map()` function in Python.
- The map() function in Python is used to apply a function to each item in an iterable (like a list or tuple) and return a new iterator with the results.

 **Purpose of map()**

 To transform data in a sequence without writing an explicit loop.
 Basic Syntax

 map(function, iterable)


```python
# Example
nums = [1, 2, 3, 4]
squares = map(lambda x: x**2, nums)

print(list(squares))  # Output: [1, 4, 9, 16]
```

10.  What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
- map(), reduce(), and filter() are built-in Python functions (or in functools) that help process data from iterables without explicit loops.

1. **map(function, iterable)**

  What it does:

 Applies a function to each item in the iterable.

 Returns a new iterable with the results.

```python
# Example
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, nums))
```

2. **filter(function, iterable)**

  What it does:

 Applies a function to each item, keeps only those for which the function returns True.

 Returns a filtered iterable.

```python
# Example
nums = [1, 2, 3, 4, 5]
evens = list(filter(lambda x: x % 2 == 0, nums))
```

3. **reduce(function, iterable)**

  What it does:

 Applies a function cumulatively, reducing the iterable to a single value.

 Must be imported from the functools module.

```python
# Example
from functools import reduce

nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)

```

```python
#  Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    if n < 2 or n > 200:
        return False  # Not prime or out of range
    for i in range(2, int(n**0.5) + 1):  # Check up to square root of n
        if n % i == 0:
            return False
    return True
```

```python
print(is_prime(2))    # True
print(is_prime(97))   # True
print(is_prime(100))  # False
print(is_prime(201))  # False (out of range)
```

```python
#  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.a
```

```python
fib = FibonacciIterator(10)

for num in fib:
    print(num)
```

```python
# Write a generator function in Python that yields the powers of 2 up to a given exponent
def powers_of_2(exponent):
    for i in range(exponent + 1):
        yield 2 ** i
```

```python
for power in powers_of_2(5):
    print(power)
```

```python
# Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()  # Remove the newline character at the end of each line
```

```python
# Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# List of temperatures in Celsius
celsius_temperatures = [0, 25, 30, 35, 40]

# Using map() to convert each Celsius temperature to Fahrenheit
fahrenheit_temperatures = list(map(celsius_to_fahrenheit, celsius_temperatures))

# Print the result
print(fahrenheit_temperatures)
```

```python
#  Create a Python program that uses `filter()` to remove all the vowels from a given string.

def remove_vowels(s):
    vowels = 'aeiouAEIOU'
    return ''.join(filter(lambda x: x not in vowels, s))

# Example string
input_string = "Hello, World!"

# Using filter() to remove vowels
output_string = remove_vowels(input_string)

# Print the result
print(output_string)
```
